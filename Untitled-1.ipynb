```python
import requests

def search_uniprot(sequence):
    url = "https://rest.uniprot.org/uniprotkb/search"
    params = {
        "query": f"sequence:{sequence}",
        "format": "json"
    }
    response = requests.get(url, params=params)
    return response.json()
```

```python
import requests
import time

def find_uniprot_id_long_sequence(sequence):
    # Bước 1: Gửi yêu cầu chạy BLAST (Dùng POST để gửi chuỗi dài)
    url = "https://rest.uniprot.org/blast/run"
    payload = {
        'ids': '', 
        'program': 'blastp', 
        'stype': 'protein', 
        'database': 'uniprotkb', 
        'sequence': sequence
    }
    
    response = requests.post(url, data=payload)
    if response.status_code != 200:
        return f"Lỗi gửi request: {response.status_code}"
    
    job_id = response.json()['jobId']
    print(f"Đang xử lý Job: {job_id}...")

    # Bước 2: Kiểm tra trạng thái đến khi xong
    status_url = f"https://rest.uniprot.org/blast/status/{job_id}"
    while True:
        status = requests.get(status_url).json().get('status')
        if status == 'FINISHED':
            break
        elif status == 'FAILURE':
            return "BLAST thất bại."
        time.sleep(3) # Đợi 3 giây rồi kiểm tra lại

    # Bước 3: Lấy kết quả UniProt ID đầu tiên
    result_url = f"https://rest.uniprot.org/blast/results/{job_id}?format=json"
    results = requests.get(result_url).json()
    
    if 'results' in results and len(results['results']) > 0:
        # Trả về UniProt ID của kết quả khớp nhất
        return results['results'][0]['entry']
    return "Không tìm thấy ID nào khớp."

# Ví dụ sử dụng
long_seq = "..." # Chuỗi dài > 500 của bạn
print(f"Kết quả: {find_uniprot_id_long_sequence(long_seq)}")
```

```python
import requests
import time
import pandas as pd

def get_uniprot_id_from_sequence(sequence):
    """
    Sử dụng UniProt BLAST API để tìm ID cho cả chuỗi ngắn và chuỗi dài (>500aa).
    """
    # 1. Khởi tạo Job BLAST (Sử dụng POST để tránh lỗi URL quá dài)
    run_url = "https://rest.uniprot.org/blast/run"
    payload = {
        'program': 'blastp',
        'stype': 'protein',
        'database': 'uniprotkb', # Có thể chọn 'uniprotkb_refprotsynch' để nhanh hơn
        'sequence': sequence
    }
    
    try:
        response = requests.post(run_url, data=payload)
        response.raise_for_status() # Kiểm tra nếu có lỗi HTTP
        job_id = response.json()['jobId']
        print(f"Đã gửi Job: {job_id}...")

        # 2. Vòng lặp chờ kết quả (Polling)
        status_url = f"https://rest.uniprot.org/blast/status/{job_id}"
        while True:
            status_res = requests.get(status_url).json()
            current_status = status_res.get('status')
            
            if current_status == 'FINISHED':
                break
            elif current_status == 'FAILURE':
                return "Lỗi: Job BLAST thất bại."
            
            time.sleep(3) # Đợi 3 giây trước khi kiểm tra lại

        # 3. Truy xuất kết quả UniProt ID
        result_url = f"https://rest.uniprot.org/blast/results/{job_id}?format=json"
        results = requests.get(result_url).json()
        
        if 'results' in results and len(results['results']) > 0:
            # Lấy ID của kết quả có độ tương đồng cao nhất (top hit)
            top_hit = results['results'][0]['entry']['primaryAccession']
            return top_hit
        else:
            return "Không tìm thấy kết quả"

    except Exception as e:
        return f"Lỗi hệ thống: {str(e)}"

# --- CÁCH ÁP DỤNG VÀO DATAFRAME CỦA BẠN ---
# Giả sử bạn muốn tìm ID cho hàng thứ 100 như trong ảnh
seq_to_find = df.loc[100, 'sequence']
uniprot_id = get_uniprot_id_from_sequence(seq_to_find)

print(f"\nKết quả cuối cùng:")
print(f"Sequence: {seq_to_find[:30]}...") # In thử 30 ký tự đầu
print(f"UniProt ID: {uniprot_id}")
```
